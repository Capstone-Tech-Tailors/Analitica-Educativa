## 3. Función de Análisis Comparativo

```python
def analizar_sentimiento_espanol(texto):
    """
    Analiza el sentimiento de un texto usando múltiples modelos para español
    """
    print(f"📝 Texto: '{texto}'")
    print("-" * 60)
    
    # Modelo Robertuito
    robertuito_resultado = modelo_robertuito(texto)[0]
    print(f"Robertuito: {robertuito_resultado['label']} (confianza: {robertuito_resultado['score']:.3f})")
    
    # Modelo BETO
    beto_resultado = modelo_beto(texto)[0]
    print(f"BETO: {beto_resultado['label']} (confianza: {beto_resultado['score']:.3f})")
    
    # Modelo Multilingüe
    multi_resultado = modelo_multilingue(texto)[0]
    print(f"Multilingüe: {multi_resultado['label']} (confianza: {multi_resultado['score']:.3f})")
    
    print("\n")
```

### 3.1 Análisis de Diferentes Tipos de Sentimientos

```python
# Oraciones con diferentes niveles de sentimiento
oraciones_variadas = [
    "¡Excelente servicio! Muy recomendable",
    "El producto está bien, nada extraordinario",
    "Terrible experiencia, no lo recomiendo para nada",
    "Me parece un producto regular, ni bueno ni malo",
    "¡Increíble! Superó todas mis expectativas",
    "Muy decepcionado con la calidad",
    "Es aceptable por el precio que tiene",
    "¡Qué maravilla! Lo volvería a comprar"
]

for oracion in oraciones_variadas:
    analizar_sentimiento_espanol(oracion)
```

## 5. Función Completa de Análisis

```python
def analisis_completo_espanol(texto, mostrar_detalles=True):
    """
    Función completa para análisis de sentimientos en español
    Devuelve el consenso de múltiples modelos
    """
    # Obtener predicciones de todos los modelos
    pred_robertuito = modelo_robertuito(texto)[0]
    pred_beto = modelo_beto(texto)[0]
    pred_multi = modelo_multilingue(texto)[0]
    
    if mostrar_detalles:
        print(f"📄 Texto: {texto}")
        print("─" * 50)
        print(f"🤖 Robertuito: {pred_robertuito['label']} ({pred_robertuito['score']:.3f})")
        print(f"🤖 BETO: {pred_beto['label']} ({pred_beto['score']:.3f})")
        print(f"🌍 Multilingüe: {pred_multi['label']} ({pred_multi['score']:.3f})")
    
    # Determinar consenso (simplificado)
    predicciones = [pred_robertuito['label'], pred_beto['label'], pred_multi['label']]
    confianzas = [pred_robertuito['score'], pred_beto['score'], pred_multi['score']]
    
    # Encontrar la predicción más confiable
    idx_max_confianza = confianzas.index(max(confianzas))
    mejor_prediccion = predicciones[idx_max_confianza]
    mejor_confianza = confianzas[idx_max_confianza]
    
    if mostrar_detalles:
        print(f"✅ Consenso: {mejor_prediccion} (confianza: {mejor_confianza:.3f})")
        print("\n")
    
    return {
        'texto': texto,
        'sentimiento': mejor_prediccion,
        'confianza': mejor_confianza,
        'detalles': {
            'robertuito': pred_robertuito,
            'beto': pred_beto,
            'multilingue': pred_multi
        }
    }
```

```python
# Ejemplo de uso de la función completa
resultado_completo = analisis_completo_espanol(
    "Este restaurante tiene una comida deliciosa pero el servicio es lento"
)

print(f"Resultado final: {resultado_completo['sentimiento']}")
```
